```python
import savana
import ee
```

```python
savana.initialize(
    project="ee-desmond"
)
```

```python
aoi = "projects/ee-desmond/assets/NewParkMerged"
```

```python
clf = savana.classify_landscape(
    aoi="projects/ee-desmond/assets/NewParkMerged",
    name_filter="Kyabobo National Park",
    epochs=[2019, 2024],
    park_name="Kyabobo",
)
```

```python
clf.show()
```

```python
clf.show_years([2019, 2024])
```

```python
clf.compare(2019, 2024)
```

```python
clf.show_change()
```
